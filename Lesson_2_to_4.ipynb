```python
#MINI DATA ANALYSIS
students=['john','mary','peter','grace','david']
scores=[65,82,48,91,73] 
#variables to track
passed_count=0
failed_count=0
#looping through both lists together
for student, score in zip(students,scores):
    #determine pass/fail status
    if score>=50:
        status="pass"
        passed_count +=1
    else:
        status="fail"
        failed_count +=1
    print(f"{student:}-{score:}-{status:}") 
#determine grade
    if score>=80:
        grade='grade A'
    elif score>=70:
        grade='grade B'
    elif score>=60:
        grade='gradeC'
    elif score>=50:
        grade='gradeD'
    else:
        grade='grade F'
    print(f"{student}-{score}-{status}-{grade}") 
   
```
