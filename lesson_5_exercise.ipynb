```python
import pandas as pd
data={
    'Product':['Laptop','Phone','Chair','Table','Headphones','Monitor'],
    'Category':['Electronics','Electronics','Furniture','Furniture','Electronics','Electronics'],
    'Price':[800,400,75,150,50,200],
    'Quantity':[3,5,10,4,12,6]
    }
df=pd.DataFrame(data) 
df 
```

```python
df['Sales']=df['Price']*df['Quantity']
sales_df=df
sales_df 
```

```python
sales_df.to_csv('sales_analysis.csv',index=False) 
```

```python
data = {
    "Name": ["Alice", "Brian", "Carol", "David"],
    "Age": [25, None, 30, 28],
    "Salary": [40000, 45000, None, 50000]
}
employees=pd.DataFrame(data)
employees 
```

```python
#Count missing values in each column
employees.isnull().sum() 
```

```python
#Check whether any values are missing
employees.isnull() 
```

```python
#Remove rows containing missing values
employees=employees.dropna()
employees 
```

```python
#Find the median age
employees['Age'].median() 
```

```python
#Fill the missing age with the median
employees=employees['Age'].fillna(employees['Age'].median())
employees 
```
