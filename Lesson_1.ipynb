```python
#PART E-CHALLENGE
#a shop records th following sales:
products=["Laptop", "Phone", "Tablet", "Keyboard"]
prices=[500,300,200,50]
quantities=[2, 5, 3, 10]
#total value of each product
for product, price, quantity in zip(products,prices,quantities):
    total= price*quantity
    print(f"{product}: {total:,}")  
#total sales value
```

```python
#total sales value
total=sum(price*quantity for price,quantity in zip(prices,quantities))
print(total) 
```
