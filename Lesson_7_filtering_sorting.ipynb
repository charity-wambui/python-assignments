```python
#Homework

#Part A: Retail Sales Analysis
#Create the following DataFrame:

sales = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet", "Monitor","Keyboard", "Mouse", "Printer", "Speaker","Camera", "Headphones"],
    "Category": ["Electronics", "Electronics", "Electronics","Electronics", "Accessories", "Accessories","Electronics", "Accessories", "Electronics","Accessories"],
    "Price": [800, 400, 300, 250, 50,25, 180, 90, 600, 120],
    "Quantity": [3, 10, 5, 7, 20,30, 4, 12, 2, 15]
})
df_sales=pd.DataFrame(sales) 
df_sales 
```

```python
#Task 1: Select data
#Display:- Products and prices.
df_sales[["Product","Price"]] 
```

```python
# Products and quantities.
df_sales[["Product","Quantity"]] 
```

```python
#Products, categories, and prices.
df_sales[["Product","Price"]] 
```

```python
#Task 2: Filter data
#Find: 1. Products costing more than 200.
df_sales[df_sales["Price"]>200] 
```

```python
#2. Products costing less than 100.
df_sales[df_sales["Price"]<100] 
```

```python
#3. Electronics products.
df_sales[df_sales["Category"]=="Electronics"] 
```

```python
#4. Accessories with quantities above 10.
df_sales[(df_sales["Category"]=="Accessories") &(df_sales["Quantity"]>10)] 
```

```python
#5. Products costing between 100 and 500.
df_sales[df_sales["Price"].between(100,500)] 
```

```python
#Task 3: Sort data
#Sort: 1. Products by price from highest to lowest.
df_sales.sort_values("Price",ascending=False) 
```

```python
#2. Products by quantity from lowest to highest.
df_sales.sort_values("Quantity") 
```

```python
#3. Electronics products by price from highest to lowest.
Electronics=df_sales[df_sales["Category"]=="Electronics"]
Electronics.sort_values("Price",ascending=False) 
```

```python
#Task 4: Calculate sales value
df_sales["sales_value"]=df_sales["Price"]*df_sales["Quantity"]
df_sales 
```

```python
#find:- The product with the highest sales value.
df_sales.loc[df_sales["sales_value"].idxmax()] 
```

```python
#- The three products with the highest sales values.
df_sales.nlargest(3,"sales_value") 
```

```python
#- Products whose sales value exceeds 1,000.
df_sales[df_sales["sales_value"]>1000] 
```

```python
#- The total sales value across all products.
df_sales["sales_value"].sum() 
```
